# Data Cleaning

```python
df.duplicated().sum()
```

```python
df.isna().sum()
```

```python
df.isna().mean() * 100
```

```python
df['CustomerID'] = df['CustomerID'].astype("string").str.strip().str.replace("C", "", regex=False)

df['CustomerID'] = pd.to_numeric(df['CustomerID'], errors="coerce")

df['CustomerID'] = df['CustomerID'].fillna(
    pd.Series(range(1, len(df) + 1), index=df.index)
)

df['CustomerID'] = ("C" + df['CustomerID'].astype(int).astype(str))
```

```python
df.head()
```

```python
df['Gender'] = df['Gender'].astype("string").str.strip().str.lower()

df['Gender'] = df['Gender'].replace({
    "m": "Male",
    "male": "Male",
    "f": "Female",
    "female": "Female"
})
```

```python
df.loc[(df['Age'] < 0) | (df['Age'] > 100), "Age"] = np.nan

df['Age'] = df['Age'].round().astype("Int64")
```

```python
df['Signup_Date'] = pd.to_datetime(
    df['Signup_Date'],
    format="%d/%m/%Y",
    errors="coerce"
)
```

```python
df['Last_purchase_date'] = pd.to_datetime(
    df['Last_purchase_date'],
    format="%d/%m/%Y",
    errors="coerce"
)
```

```python
df.loc[df['purchase_amount'] < 0, "purchase_amount"] = np.nan
```

```python
df.loc[~df['feedback_score'].between(0, 10), "feedback_score"] = np.nan
```

```python
email_pattern = r"^[^@\s]+@[^@\s]+\.[^@\s]+$"

valid_email = df["email"].astype("string").str.match(
    email_pattern,
    na = False
)

df.loc[~valid_email, "email"] = np.nan
```

```python
df["Phone_number"] = df["Phone_number"].astype("string").str.strip()

valid_phone = df["Phone_number"].str.fullmatch(
    r"\d{10}",
    na=False
)

df.loc[~valid_phone, "Phone_number"] = np.nan
```

```python
df.to_csv(
    "cleaned_date.csv",
    index=False
)
```
